# Dataset Documentation

## Overview

This dataset contains country-level information about population, CO₂ emissions, GDP, geographic coordinates, income classification, and lending category. It contains **5,544 rows** and **13 columns**.

## Original Columns

| Column | Description |
|---|---|
| `iso2c` | Two-letter country or region code |
| `country` | Country or regional aggregate name |
| `year` | Year of the observation |
| `SP.POP.TOTL` | Total population |
| `EN.ATM.CO2E.PC` | CO₂ emissions per capita |
| `NY.GDP.PCAP.KD` | GDP per capita in constant 2015 US dollars |
| `iso3c` | Three-letter country code |
| `region` | Geographical region |
| `capital` | Capital city |
| `longitude` | Longitude of the capital or location |
| `latitude` | Latitude of the capital or location |
| `income` | World Bank income classification |
| `lending` | World Bank lending category |

## Derived Variables

| Variable | Description |
|---|---|
| `co2_emissions` | NumPy array containing non-null CO₂ emission values |
| `scores` | Non-null CO₂ emission values used for analysis |
| `income` | Currently contains CO₂ emission values despite its name |
| `rating_category` | Category assigned to observations based on calculated scores |
| `result` | Classification label such as `Low` |
| `outliers` | CO₂ emission values outside the calculated IQR limits |
| `Q1` | First quartile: approximately `0.748` |
| `Q3` | Third quartile: approximately `6.766` |
| `IQR` | Interquartile range: approximately `6.019` |
| `lower_limit` | Lower outlier boundary: approximately `-8.280` |
| `upper_limit` | Upper outlier boundary: approximately `15.794` |
| `mask` | Boolean array identifying outlier records |
| `high_rated` | DataFrame intended to contain high-rated countries; currently empty |
| `movies` | DataFrame containing population, GDP per capita, and latitude; the name is unrelated to the dataset |

## Data Quality Notes

- Missing values exist in CO₂ emissions, capital, latitude, and longitude.
- `AudienceScore` is currently `None` and is not part of this dataset.
- The variable `income` contains CO₂ values and should be renamed to avoid confusion.
- The variable `movies` should be renamed because this is country-level economic and environmental data.
- The outlier analysis is based on CO₂ emissions per capita.

### 📊 Interquartile Range (IQR)

This code calculates the **Interquartile Range (IQR)** of the `income` data.

* **Q1 (25th Percentile):** The value below which 25% of the data falls.
* **Q3 (75th Percentile):** The value below which 75% of the data falls.
* **IQR:** The difference between Q3 and Q1.

**Formula:**

`IQR = Q3 - Q1`

### Why do we use IQR?

IQR helps us understand the **spread of the middle 50% of the data**. It is also useful for identifying **outliers**.

A common outlier rule is:

* Lower Bound = Q1 − 1.5 × IQR
* Upper Bound = Q3 + 1.5 × IQR

Values outside these bounds can be considered potential outliers.


````python
Q1 = np.percentile(income, 25)
Q3 = np.percentile(income, 75)
IQR = Q3 - Q1
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
````

### 📌 Outlier Detection Using IQR

This code identifies **outliers** in the `income` data using the **IQR (Interquartile Range) method**.

#### Steps:

1. **Remove missing values**

   * `np.isnan()` is used to identify missing values.
   * Only valid numerical values are kept in `scores`.

2. **Calculate Q1 and Q3**

   * **Q1 (25th Percentile):** 25% of the data is below this value.
   * **Q3 (75th Percentile):** 75% of the data is below this value.

3. **Calculate IQR**

   `IQR = Q3 - Q1`

4. **Calculate Limits**

   **Lower Limit:**

   `Q1 - 1.5 × IQR`

   **Upper Limit:**

   `Q3 + 1.5 × IQR`

5. **Find Outliers**

   * Values below the lower limit or above the upper limit are considered **outliers**.

6. **Count Outliers**

   * `len(outliers)` gives the total number of outliers.

### 🎯 Purpose

This analysis helps identify **unusually high or low values** in the dataset and is useful for data cleaning and preprocessing before statistical analysis or machine learning.

**Note:** Your code prints the same results twice at the end. One set of `print()` statements is enough.


````python
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

scores = income[~np.isnan(income)]
Q1 = np.percentile(scores, 25)
Q3 = np.percentile(scores, 75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = scores[(scores < lower_limit) | (scores > upper_limit)]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:", outliers)
print("Number of Outliers:", len(outliers))

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Outliers:", outliers)
print("Number of Outliers:", len(outliers))
````
